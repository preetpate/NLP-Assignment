```python
# 2.Use the Naive Bayes algorithm from scikit-learn to train a text classifier that predicts whether a Zomato review is positive 
# or negative, using a small sample dataset of at least 20 reviews.

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
reviews = [
    "The food was delicious",
    "Amazing taste and excellent service",
    "I loved the pizza",
    "The restaurant was fantastic",
    "Very tasty food",
    "Great ambience and friendly staff",
    "Excellent experience",
    "The food was fresh and delicious",
    "Really enjoyed the meal",
    "Amazing restaurant and great food",

    "The food was terrible",
    "Very bad experience",
    "I hated the food",
    "The service was horrible",
    "Food was cold and tasteless",
    "Very disappointing restaurant",
    "The pizza was awful",
    "Bad food and poor service",
    "Worst restaurant experience",
    "I will never visit again"
]

labels = [
    "positive", "positive", "positive", "positive", "positive",
    "positive", "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative", "negative",
    "negative", "negative", "negative", "negative", "negative"
]

X_train, X_test, y_train, y_test = train_test_split(
    reviews,
    labels,
    test_size=0.3,
    random_state=42,
    stratify=labels
)
```

```python
vectorizer = CountVectorizer()

X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

model = MultinomialNB()

model.fit(X_train_vectorized, y_train)

predictions = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, predictions)

print("Naive Bayes Accuracy:", accuracy)
```

```python
# 3.Build a spam detector for WhatsApp-style messages using Logistic Regression from scikit-learn — train on at least 15 example 
# messages labeled as 'spam' or 'not spam', then predict the label for 3 new custom messages of your choice.

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
messages = [
    "Congratulations you won a free prize",
    "You have won a lottery ticket",
    "Click this link to claim your free reward",
    "Win exciting cash prizes now",
    "You are selected for a free gift",
    "Get free recharge by clicking this link",
    "Congratulations claim your reward now",
    "You won a free shopping voucher",

    "Hey, are you coming to class today",
    "Please send me the assignment",
    "Let's meet at 5 pm",
    "Can you call me later",
    "Your order has been delivered",
    "I will reach home soon",
    "Please bring your notebook",
    "See you tomorrow"
]

labels = [
    "spam", "spam", "spam", "spam",
    "spam", "spam", "spam", "spam",
    "not spam", "not spam", "not spam", "not spam",
    "not spam", "not spam", "not spam", "not spam"
]

X_train, X_test, y_train, y_test = train_test_split(
    messages,
    labels,
    test_size=0.25,
    random_state=42,
    stratify=labels
)
```

```python
vectorizer = TfidfVectorizer()

X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

model = LogisticRegression()

model.fit(X_train_vectorized, y_train)

predictions = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, predictions)

print("Logistic Regression Accuracy:", accuracy)
```

```python
new_messages = [
    "Congratulations you won a free gift",
    "Please send me today's homework",
    "Click now to claim your cash reward"
]

new_messages_vectorized = vectorizer.transform(new_messages)

new_predictions = model.predict(new_messages_vectorized)

print("\nNew Message Predictions:")

for message, prediction in zip(new_messages, new_predictions):
    print(message, "->", prediction)
```

```python
# 4.Compare the accuracy of Naive Bayes and Logistic Regression models on your Zomato review sentiment dataset and print 
# which model performs better.<br><br><em><strong>Hint:</strong> Use train_test_split to create training and testing sets, and accuracy_score to evaluate.</em>

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
reviews = [
    "The food was delicious",
    "Amazing taste and excellent service",
    "I loved the pizza",
    "The restaurant was fantastic",
    "Very tasty food",
    "Great ambience and friendly staff",
    "Excellent experience",
    "The food was fresh and delicious",
    "Really enjoyed the meal",
    "Amazing restaurant and great food",

    "The food was terrible",
    "Very bad experience",
    "I hated the food",
    "The service was horrible",
    "Food was cold and tasteless",
    "Very disappointing restaurant",
    "The pizza was awful",
    "Bad food and poor service",
    "Worst restaurant experience",
    "I will never visit again"
]

labels = [
    "positive", "positive", "positive", "positive", "positive",
    "positive", "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative", "negative",
    "negative", "negative", "negative", "negative", "negative"
]

X_train, X_test, y_train, y_test = train_test_split(
    reviews,
    labels,
    test_size=0.3,
    random_state=42,
    stratify=labels
)
```

```python
vectorizer = TfidfVectorizer()

X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)

nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)

nb_predictions = nb_model.predict(X_test)
nb_accuracy = accuracy_score(y_test, nb_predictions)

lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)

lr_predictions = lr_model.predict(X_test)
lr_accuracy = accuracy_score(y_test, lr_predictions)
```

```python
print("Naive Bayes Accuracy:", nb_accuracy)
print("Logistic Regression Accuracy:", lr_accuracy)

if nb_accuracy > lr_accuracy:
    print("Naive Bayes performs better.")
elif lr_accuracy > nb_accuracy:
    print("Logistic Regression performs better.")
else:
    print("Both models have the same accuracy.")
```
