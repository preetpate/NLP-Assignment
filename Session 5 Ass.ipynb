```python
#2.Write a short paragraph (2-3 sentences) explaining one limitation of traditional NLP models (like Bag of Words or TF-IDF)
# when trying to understand the meaning of a WhatsApp chat message.

Traditional NLP techniques such as Bag of Words and TF-IDF mainly represent text using individual word frequencies and do not properly
understand the context of a sentence. For example, in a WhatsApp message, the meaning of a word can change depending on the surrounding
words, but traditional models have limited ability to understand this context.
```

```python
#3.Use the Hugging Face pipeline to perform sentiment analysis on three recent Zomato or Swiggy app reviews 
# (copy-paste the review text) and print the predicted sentiment for each.

from transformers import pipeline

sentiment_analyzer = pipeline("sentiment-analysis")

reviews = [
    "The food was delicious and the delivery was very fast!",
    "The food was cold and the service was very disappointing.",
    "The pizza was good but the delivery was a little slow."
]

for review in reviews:
    result = sentiment_analyzer(review)

    print("Review:", review)
    print("Sentiment:", result[0]["label"])
    print("Confidence:", round(result[0]["score"], 4))
    print()
```

```python
# 5.Use ChatGPT or Copilot to generate a simple Python function that demonstrates the attention mechanism with three short
# sentences as input, and explain in your own words (2-3 lines) what the attention scores mean in your output.<br><br><em><strong>
# Hint:</strong> Ask the AI tool to 'show a basic attention calculation example in Python.'</em>

import numpy as np

def softmax(scores):
    exp_scores = np.exp(scores - np.max(scores))
    return exp_scores / np.sum(exp_scores)


def attention(sentence):
 
    words = sentence.split()
    
    vectors = np.array([
        [1.0, 0.0],
        [0.0, 1.0],
        [1.0, 1.0]
    ])

    scores = np.dot(vectors, vectors.T)

    attention_scores = softmax(scores[0])

    print("Sentence:", sentence)
    print("Words:", words)
    print("Attention Scores:", attention_scores)
```

```python
sentences = [
    "I love pizza",
    "The food is delicious",
    "Delivery was very fast"
]

for sentence in sentences:
    attention(sentence)
    print()
```
