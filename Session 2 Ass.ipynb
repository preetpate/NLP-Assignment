```python
# 2.Create a function to remove all punctuation marks from a Flipkart product review string and return the cleaned text.
# <br><br><em><strong>Hint:</strong> Use the string.punctuation constant from the Python string module.</em>

import string

def remove_punctuation(review):
    cleaned_text = ""

    for char in review:
        if char not in string.punctuation:
            cleaned_text += char

    return cleaned_text
```

```python
review = input("Enter your Flipkart product review: ")

cleaned_review = remove_punctuation(review)

print("Cleaned Review:", cleaned_review)
```

```python
# 4.Download a list of English stopwords using NLTK and write a function that removes all stopwords from a WhatsApp message string.

import nltk

nltk.download('stopwords')

from nltk.corpus import stopwords
```

```python
def remove_stopwords(message):
    stop_words = set(stopwords.words('english'))

    words = message.split()

    filtered_words = []

    for word in words:
        if word.lower() not in stop_words:
            filtered_words.append(word)

    return " ".join(filtered_words)
```

```python
message = input("Enter your WhatsApp message: ")

cleaned_message = remove_stopwords(message)

print("Original Message:", message)
print("Message without Stopwords:", cleaned_message)
```

```python
# 5.Use NLTK’s PorterStemmer to stem each word in a list of trending hashtags from Twitter, and print both the original and stemmed 
# versions side by side.<br><br><em><strong>Hint:</strong> Example hashtags: ['#playing', '#dancing', '#eating']</em>

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

hashtags = ['#playing', '#dancing', '#eating']

print("Original Hashtag\tStemmed Word")

for hashtag in hashtags:
    word = hashtag.replace("#", "")
    stemmed_word = stemmer.stem(word)

    print(f"{hashtag}\t\t{stemmed_word}")
```
