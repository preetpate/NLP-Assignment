```python
# 1.Take three recent Instagram captions (your own or your friends'), and manually create a vocabulary list of all unique words used,
# ignoring case and punctuation.

import string

captions = [
    "Love exploring new places with friends!",
    "Amazing food and beautiful places.",
    "Friends, food and travel make life happy!"
]

vocabulary = set()
```

```python
for caption in captions:
    caption = caption.lower()

    caption = caption.translate(str.maketrans("", "", string.punctuation))

    words = caption.split()

    vocabulary.update(words)

vocabulary = sorted(vocabulary)

print("Vocabulary:")
print(vocabulary)
```

```python
#2.Write a Python script that converts a list of short Zomato food reviews (e.g., 'Loved the pizza', 'Service was slow', 'Great ambience') 
# into a Bag of Words matrix using scikit-learn's CountVectorizer, and print the resulting matrix.

from sklearn.feature_extraction.text import CountVectorizer

reviews = [
    "Loved the pizza",
    "Service was slow",
    "Great ambience"
]

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(reviews)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
#3.Given an array of WhatsApp chat messages, count and display the frequency of each word that appears across all messages.
# <br><br><em><strong>Hint:</strong> Use a dictionary to store word counts and ignore common stopwords like 'the', 'is', 'at'.</em>

messages = [
    "I am at the market",
    "The market is crowded",
    "I am going to the market"
]

stopwords = {"the", "is", "at"}

word_count = {}
```

```python
for message in messages:
    words = message.lower().split()

    for word in words:
        if word not in stopwords:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1

print("Word Frequency:")

for word, count in word_count.items():
    print(word, ":", count)
```

```python
#4.Use scikit-learn's TfidfVectorizer to transform a list of Flipkart product reviews into TF-IDF vectors, then print the TF-IDF 
# value for the word 'quality' in each review.<br><br><em><strong>Constraint:</strong> Use at least 3 reviews and show the output for each.</em>

from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    "The product has excellent quality",
    "Good quality product and fast delivery",
    "The quality of this product is amazing"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(reviews)

feature_names = vectorizer.get_feature_names_out()

quality_index = list(feature_names).index("quality")

print("TF-IDF value for 'quality':")

for i, review in enumerate(reviews):
    value = tfidf_matrix[i, quality_index]
    print(f"Review {i + 1}: {value:.4f}")
```

```python
#5.In your own words, list two key differences between Bag of Words and TF-IDF, and give a real-world example
# (from any app you use) where TF-IDF would be more useful than Bag of Words.

Bag of Words
Gives importance mainly based on how many times a word appears.
Common words can get high values if they occur frequently.

TF-IDF
Gives higher importance to words that are important in a document but less common across documents.
Common words across many documents get lower weights.
```
