```python
# 4.Build a simple sentiment checker in Python: ask the user to enter a review about a movie (like on BookMyShow),
# then check if the review contains the word 'good' or 'bad' and print 'Positive', 'Negative', or 'Neutral'.<br><br><em><strong>Constraint:
# </strong> Only use basic Python (no external libraries).</em>

review = input("Enter your movie review: ")

review = review.lower()

if "good" in review:
    print("Positive")
elif "bad" in review:
    print("Negative")
else:
    print("Neutral")
```

```python
# 5.Imagine you are designing a spam filter for an email app. Write a Python function is_spam(email_text) that returns True if the
# email_text contains the word 'win' or 'free', otherwise returns False. Test your function with two examples.

def is_spam(email_text):
    
    email_text = email_text.lower()
    
    if "win" in email_text or "free" in email_text:
        return True
    
    return False
```

```python
email1 = "Congratulations! You can win a free prize."

print(is_spam(email1))
```

```python
email2 = "Your order has been shipped successfully."

print(is_spam(email2))
```
